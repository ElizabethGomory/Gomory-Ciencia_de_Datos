Análisis de Sentimientos.** El siguiente análisis de sentimientos  permitirá enseñarle a una computadora a leer emociones en los textos que las personas escriben, en nuestro caso, redes sociales y web. A través de esta técnica, podemos clasificar automáticamente los comentarios como positivos (“bueno”), negativos (“malo”) o neutros/informativos (“info”), lo que nos permitirá entender cómo se sienten los clientes sin tnecesidad de leer uno por uno, cada comentario. **1.b.** Procesamiento del Análisis de Sentimientos

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(10, 8))
ax.axis('off')
steps = [
    "Inicio",
    "Importación de Datos\n- Librerías (pandas, re)\n- Lectura del CSV",
    "Exploración Inicial\n- df.head()\n- df.info()",
    "Análisis Exploratorio\n- Distribución por tipo de comentario",
    "Preprocesamiento de Texto\n- Tokenización con regex",
    "Definición de Variables\n- X: Comentarios\n- y: Etiquetas",
    "Fin: Datos listos para modelado"]

y_positions = list(reversed(range(len(steps))))
for i, (step, y) in enumerate(zip(steps, y_positions)):
    ax.text(0.5, y, step, ha='center', va='center',
            bbox=dict(boxstyle="round,pad=0.5", fc="lightblue", ec="navy", lw=1.5),
            fontsize=10)
    if i < len(steps) - 1.8:
        ax.annotate('', xy=(0.5, y - 0.5), xytext=(0.5, y - 0.1),
                    arrowprops=dict(arrowstyle="->", color='gray', lw=1.5))
ax.set_ylim(-0.5, len(steps))
ax.set_xlim(0, 1)
plt.tight_layout()
plt.show()
```
