#### Importación de librerías y archivo ".xlsx"

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df_titanic = pd.read_excel('/content/titanic_M1.xlsx')
```

### **2.** **Columna pclass:** completar valores NA = asignación de "clase 2"

```python
titanic = df_titanic
titanic['pclass']=titanic['pclass'].fillna(2.0)
titanic
```

### **3.** **Columna sex:** reemplazar valores por "male" y "female"

```python
titanic['sex'] = titanic['sex'].replace('Hombre', 'male')
titanic['sex'] = titanic['sex'].replace('Mujer', 'female')
titanic.head(30)
```


### **4.** **Columna edad:** completar valores NA = asignación mean




```python
mean= titanic['age'].mean()
mean
```

```python
titanic['age']=titanic['age'].fillna(mean)
titanic.head(30)
```

### **5.** **Columna embarked:** completar valores NA = asignación "S"

```python
titanic['embarked']=titanic['embarked'].fillna('S')
titanic.head(30)
```
