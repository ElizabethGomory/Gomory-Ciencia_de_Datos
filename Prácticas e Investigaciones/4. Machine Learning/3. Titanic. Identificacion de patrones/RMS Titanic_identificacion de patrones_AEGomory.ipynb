# B. Completar valores Faltantes en bases de datos.

`No se reemplazarán los valores de "Cabin" en ambas bases de datos, ya que los valores nulos son mayores al 75%, por lo que se decidió no considerarlos en el modelo.`

```python
test.info()
```

```python
train.info()
```

```python
pre_columns=["Sex", "Embarked"]
for c in pre_columns:
    encoder = LabelEncoder()
    train [c] = encoder.fit_transform(train [c]. astype("str"))
    test [c] = encoder.fit_transform(test [c].astype("str"))

train.info()
test.info()
```

```python
x_columns = ['Age', 'Embarked', 'Fare']
imputer = KNNImputer (n_neighbors=3,weights= "uniform")
train[x_columns] = imputer.fit_transform(train[x_columns])
test[x_columns] = imputer.fit_transform(test[x_columns])
```

```python
display(test[x_columns].info())
print(test.isnull().sum())
display(train[x_columns].info())
print(train.isnull().sum())
```

# C. Ajustar Modelo de Árbol de Decisión.

`Se decidió trabajar con las columnas de: Pclass, Sex, Age, SibSp, Parch, Faree y Embarked; para el entrenamiento, el resto de las columnas continen valores únicos que resultarían en un subajuste.`

```python
col = ['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare', 'Embarked']

label_col = 'Survived'
seed_num = 10
x, y = train[col], train[label_col]

print(x.columns)
```

```python
x_train, x_val, y_train, y_val = tts (x, y, test_size=0.3, random_state=seed_num)
```

```python
ard = DecisionTreeClassifier (random_state=seed_num)
ard.fit (x_train, y_train)
```

#  D. Entrenamiento del Modelo

```python
y_pred = ard.predict(x_val)
acc = accs(y_val, y_pred)
pre = prs (y_val, y_pred)
rec = recs (y_val, y_pred)
f1 = f1s (y_val, y_pred)

print ("Accuracy: ", acc)
print ("Precision: ", pre)
print ("Recall: ", rec)
print ("F1: ", f1)
```
