Reordenar datos NaN de mayor a menor

```python
missing_data = df.isnull().sum().sort_values(ascending=False)
missing_data = missing_data[missing_data > 0]  
missing_data
```

Limpieza de datos

```python
# Decidí eliminar solo las columnas con más del 50% de valores nulos, de lo contrario, me quedaría sin datos a entrenar
threshold = 0.5 * len(df)
columns_drop = missing_data[missing_data > threshold].index.tolist()
df_cleaned = df.drop(columns=columns_drop)
```

```python
df_cleaned = df_cleaned.dropna()
df_cleaned.shape
```

## Entrenamiento y Validación

```python
X = df_cleaned.drop("SalePrice", axis=1)
y = df_cleaned["SalePrice"]

X_encoded = pd.get_dummies(X, drop_first=True)

X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=42)
```

# Regresión Lineal

```python
model = LinearRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

comparasion = pd.DataFrame({
    "Precio Real": y_test.values[:5],
    "Predicción": y_pred[:5].round(2)
})
comparasion
```

* Estimación del valor del modelo de Regresión - MSE

```python
mse = mean_squared_error(y_test, y_pred)

print("MSE con Regresión Lineal:", mse)
```
