```python
# Divisible 3 = Fizz
# Divisible 5 = Buzz
# Divisible por 3 y 5 = FizzBuzz
# No Divisible por 3 o 5 = se imprime el número

for num in range(1,21):
    if num%3 == 0 and num%5 ==0:
        print(f"Fizz Buzz")
    elif num%3 == 0:
        print(f"Fizz")
    elif num%5 ==0:
         print(f"Buzz")
    else:
          print(num)  
```
